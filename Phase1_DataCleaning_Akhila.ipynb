```python
ratings=pd.read_csv("ratings.csv")
reviews=pd.read_csv("reviews.csv")
subscribers=pd.read_csv("subscribers.csv")
titles=pd.read_csv("titles.csv")
watch_history=pd.read_csv("watch_history.csv")
watchlist=pd.read_csv("watchlist.csv")                          
```

```python
tables = {
    "subscribers": subscribers, "titles": titles, "watch_history": watch_history,
    "ratings": ratings, "reviews": reviews, "watchlist": watchlist
}

for name, df in tables.items():
    print(f"===== {name} =====")
    print("Rows:", df.shape[0], "| Columns:", df.shape[1])
    print(df.dtypes)
    print()
```

```python
for name, df in tables.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    print(f"===== {name} =====")
    if len(missing) == 0:
        print("No missing values")
    else:
        pct = (missing / len(df) * 100).round(1)
        print(pd.DataFrame({"missing_count": missing, "missing_%": pct}))
    print()
```

```python
for name, df in tables.items():
    print(name, "→ duplicate rows:", df.duplicated().sum())

print()
print("Duplicate subscriber_id:", subscribers["subscriber_id"].duplicated().sum())
print("Duplicate title_id:", titles["title_id"].duplicated().sum())
print("Duplicate watch_id:", watch_history["watch_id"].duplicated().sum())
```

```python
subscribers["signup_date"]   = pd.to_datetime(subscribers["signup_date"])
subscribers["churn_date"]    = pd.to_datetime(subscribers["churn_date"])
titles["date_added"]         = pd.to_datetime(titles["date_added"])
titles["license_expiry"]     = pd.to_datetime(titles["license_expiry"])
watch_history["watch_date"]  = pd.to_datetime(watch_history["watch_date"])
ratings["rating_date"]       = pd.to_datetime(ratings["rating_date"])
reviews["review_date"]       = pd.to_datetime(reviews["review_date"])
watchlist["added_date"]      = pd.to_datetime(watchlist["added_date"])

print(subscribers.dtypes)
```

```python
churned = subscribers[subscribers["is_active"] == False]
bad_churn = churned[churned["churn_date"] <= churned["signup_date"]]
print("Churned customers:", len(churned))
print("Churn date before/same as signup (problem):", len(bad_churn))
```

```python
active_with_churn = subscribers[(subscribers["is_active"] == True) &
                                (subscribers["churn_date"].notnull())]
print("Active but has churn_date (problem):", len(active_with_churn))

churned_no_date = subscribers[(subscribers["is_active"] == False) &
                              (subscribers["churn_date"].isnull())]
print("Churned but no churn_date (problem):", len(churned_no_date))
```

```python
too_long = watch_history[watch_history["watch_duration_min"] >
                         watch_history["content_duration_min"]]
print("Sessions longer than the content:", len(too_long))
print("As % of all sessions:", round(len(too_long) / len(watch_history) * 100, 2))
```

```python
expected = watch_history["watch_duration_min"] / watch_history["content_duration_min"] * 100
difference = (watch_history["completion_pct"] - expected).abs()
print("Rows where completion_pct is off by more than 1 point:", (difference > 1).sum())
```

```python
print(reviews["sentiment"].value_counts(dropna=False))
```

```python
def check_ids(child, child_name, column, parent):
    missing = ~child[column].isin(parent[column])
    print(f"{child_name}.{column} not found in parent table: {missing.sum()}")

for df, name in [(watch_history, "watch_history"), (ratings, "ratings"),
                 (reviews, "reviews"), (watchlist, "watchlist")]:
    check_ids(df, name, "subscriber_id", subscribers)
    check_ids(df, name, "title_id", titles)
```

```python
short_active = subscribers[(subscribers["tenure_months"] < 3) &
                           (subscribers["is_active"] == True)]
print("Active subscribers with under 3 months tenure:", len(short_active))
```

```python
subscribers.to_csv("clean_subscribers.csv", index=False)
titles.to_csv("clean_titles.csv", index=False)
watch_history.to_csv("clean_watch_history.csv", index=False)
ratings.to_csv("clean_ratings.csv", index=False)
reviews.to_csv("clean_reviews.csv", index=False)
watchlist.to_csv("clean_watchlist.csv", index=False)
print("Clean files saved!")
```

## 1. Overview
- 6 files checked: subscribers, titles, watch_history, ratings, reviews, watchlist
- Tool used: Python (pandas) in Jupyter Notebook
- About 979,000 rows in total; the biggest table is watch_history (650,000 sessions)
- All row counts match the project brief, so nothing was lost when loading

| File | Rows | Columns |

| watch_history.csv | 650,000 | 10 |
| ratings.csv | 130,000 | 5 |
| reviews.csv | 110,000 | 7 |
| watchlist.csv | 65,000 | 5 |

## 2. Missing Values
- Only 2 columns had missing values; the other 4 tables had none
- `churn_date`: 11,199 missing (74.7%)
  - Expected: 11,199 is exactly the number of active subscribers, who have not left, so they have no churn date
- `license_expiry`: 1,966 missing (21.8%)
  - Expected: all 1,966 are StreamFlix Originals, which StreamFlix owns, so there is no licence to expire
- Decision: kept as they are, because these blanks are correct and not errors

## 3. Duplicates
- 0 duplicate rows in all 6 tables
- Unique IDs checked: `subscriber_id`, `title_id`, `watch_id`, all with 0 duplicates
- Every record is unique

## 4. Data Types
- 8 date columns were loaded as text: `signup_date`, `churn_date`, `date_added`, `license_expiry`, `watch_date`, `rating_date`, `review_date`, `added_date`
- Converted them to real dates using `pd.to_datetime()`
- Reason: so dates can be compared and grouped by month in later phases
- Numeric and True/False columns were already the correct type

## 5. Validation Checks
- **Churn vs signup date:** 0 of 3,801 churned subscribers have a churn date on or before their signup date
- **Active vs churn date:** 0 active subscribers have a churn date, and 0 churned subscribers are missing one
- **Watch vs content duration:** 0 sessions are longer than the content itself
- **Completion % formula:** 0 rows differ by more than 1 point from watch ÷ content × 100
- **Sentiment values:** only Positive (75,007), Neutral (22,759) and Negative (12,234); no invalid values
- **Orphan IDs:** 0 in all 8 checks; every subscriber_id and title_id in watch_history, ratings, reviews and watchlist exists in its parent table
- **Short tenure:** 486 active subscribers have under 3 months of tenure; these are new customers, so this is fine and only noted

## 6. Other Observations
- `license_type` has 4 values: Original, Exclusive License, Non-Exclusive License and Co-Production (2,396 titles)
- The SQL schema documents only the first 3
- Co-Production appears to be a valid category missing from the documentation, so it was kept

## 7. Actions Taken
- **Changed:** converted 8 date columns from text to dates
- **Removed:** nothing
- **Why nothing was removed:** there were no duplicates, impossible values or broken links, and the missing values are expected; removing rows would have lost correct data
- **Saved:** clean files as `clean_subscribers.csv`, `clean_titles.csv`, `clean_watch_history.csv`, `clean_ratings.csv`, `clean_reviews.csv` and `clean_watchlist.csv`

## 8. Conclusion
- The data is clean and complete; every check passed
- The only blanks are expected and explained
- High confidence in the data; it is ready for Phase 2 (charts) and Phase 3 (KPIs)
